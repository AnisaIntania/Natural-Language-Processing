```python
import spacy
import textacy
from textacy import extract

nlp = spacy.load("en_core_web_sm")
print("Model berhasil dimuat!")
```

```python
#Load a spacy model, which will be used for all further processing.
en = textacy.load_spacy_lang("en_core_web_sm")
```

```python
#Let us use a sample text file, nlphistory.txt, which is the text from the history section of Wikipedia's
#page on Natural Language Processing
#https://en.wikipedia.org/wiki/Natural_language_processing
try :
    from google.colab import files
    uploaded=files.upload()
    mytext = open('nlphistory.txt').read()

except ModuleNotFoundError :
    mytext = open('Data/nlphistory.txt').read()
```

```python
doc = textacy.make_spacy_doc(mytext, lang=en)
```

```python
import networkx as nx

# Textacy 0.12.0 masih memakai pagerank_scipy,
# sedangkan NetworkX versi baru sudah menghapusnya.
nx.pagerank_scipy = nx.pagerank
```

```python
textrank = textacy.extract.keyterms.textrank(doc, topn=5)
print(textrank)
```

```python
# Print the keywords using TextRank algorithm
textrank = textacy.extract.keyterms.textrank(
    doc,
    normalize="lemma",
    topn=5
)
print("Textrank output:", [kps for kps, weights in textrank])
# Print the keywords and phrases using SGRank algorithm
sgrank = textacy.extract.keyterms.sgrank(
    doc,
    topn=5
)
print("SGRank output:", [kps for kps, weights in sgrank])
```

```python
# Mengambil key phrases menggunakan SGRank
sgrank = textacy.extract.keyterms.sgrank(doc)

# Mengambil nama term saja
terms = set([term for term, weight in sgrank])

print("Terms:")
print(terms)
```

```python
# Mengambil semua noun chunks dari dokumen
noun_chunks = list(textacy.extract.noun_chunks(doc))
print(noun_chunks)
```
